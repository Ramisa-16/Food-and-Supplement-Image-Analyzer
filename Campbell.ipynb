````python
import requests
from PIL import Image
import google.generativeai as genai
from google.cloud import vision
from google.cloud.vision_v1 import types

# Configure your API key for genai
api_key = "AIzaSyB01ouUyKFz7IGzCgEnBdHnEY9QRQXPfIQ"
genai.configure(api_key=api_key)

# Set up Google Vision API client
vision_client = vision.ImageAnnotatorClient()

def get_image_description(image_path):
    """Uses Google Vision API to detect labels in an image and return a description."""
    with open(image_path, 'rb') as image_file:
        content = image_file.read()

    image = types.Image(content=content)
    response = vision_client.label_detection(image=image)
    labels = response.label_annotations

    # Create a simple text description from the detected labels
    descriptions = [label.description for label in labels]
    return ", ".join(descriptions)

def generate_food_name(image_path):
    """Generate food name using GenAI based on image description."""
    # Step 1: Get the image description from the Vision API
    image_description = get_image_description(image_path)

    # Step 2: Use the description to prompt the GenAI model for food recognition
    prompt = f"Based on this image description: {image_description}, what food item is it?"

    # Step 3: Call the Generative AI model for generating food name
    model = genai.GenerativeModel("gemini-1.5-flash")
    response = model.generate_content([prompt])

    # Step 4: Return the result (the name of the food)
    return response.text.strip()

# Example usage
image_path = "/content/Screenshot 2025-02-21 141509.png"  # Path to the image
food_name = generate_food_name(image_path)
print(food_name)
````
